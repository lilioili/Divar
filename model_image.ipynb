```python
import pandas as pd
from sklearn.metrics import confusion_matrix
import seaborn as sns
import  matplotlib.pyplot  as plt 

data_matrix=pd.read_excel('model_c_prediction.xlsx')

predictions=data_matrix["prediction"].values
truth=data_matrix["truth"].values

cm=confusion_matrix(truth,predictions,labels=["re","bicycle","sofa","Tv","laptop"])

plt.figure(figsize=(5,6))
sns.heatmap(cm,annot=True,xticklabels=["re","bicycle","sofa","Tv","laptop"],yticklabels=["re","bicycle","sofa","Tv","laptop"])

plt.show()
```

```python
from sklearn.metrics import classification_report

print(classification_report(truth,predictions))
```
